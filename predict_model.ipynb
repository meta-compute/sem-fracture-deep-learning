# Predict fracture classes

Select an existing checkpoint and split below. Validation is the default; switch to `test` only for final evaluation. This notebook does not train a model or modify checkpoints.

```python
from pathlib import Path

import pandas as pd
import torch

from sem_fracture.pipeline import run_prediction

DATA_DIR = Path.cwd()
MODEL_KEY = "tinyvit_21m"  # resnet18, deit_tiny, tinyvit_11m, tinyvit_21m
MODE = "finetune"  # head_only or finetune
CHECKPOINT_PATH = (
    DATA_DIR / "results" / "shared_pipeline"
    / f"{MODEL_KEY}_{MODE}_best.pth"
)
PREDICTION_SPLIT = "validation"  # change to test for final evaluation
DEVICE = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
```

```python
result = run_prediction(
    data_dir=DATA_DIR,
    model_key=MODEL_KEY,
    checkpoint_path=CHECKPOINT_PATH,
    split=PREDICTION_SPLIT,
    split_file="dataset_split.csv",
    device=DEVICE,
    batch_size=16,
)

print(result["metrics"])
pd.DataFrame(result["predictions"])
```
